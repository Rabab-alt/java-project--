```python
# ======================================================
# Crop Disease Detection Using Traditional ML Features
# Symptom-based (Non-image) Dataset
# ======================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# ------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------
try:
    df = pd.read_csv("data.csv")
    print("Dataset Loaded Successfully\n")
    print(df.head())
    print("\nDataset Info:")
    print(df.info())
except FileNotFoundError:
    print("❌ Error: 'data.csv' not found. Please upload the file.")

# ------------------------------------------------------
# 2. IDENTIFY TARGET COLUMN
# ------------------------------------------------------
# CHANGE THIS if your column name is different
TARGET_COLUMN = "Disease_Present"

# ------------------------------------------------------
# 3. DATA PREPROCESSING
# ------------------------------------------------------
print("\n--- Preprocessing Data ---")

# Convert Yes/No to 1/0
df = df.replace({"Yes": 1, "No": 0, "yes": 1, "no": 0})

# Separate features and label
X = df.drop(columns=[TARGET_COLUMN])
y = df[TARGET_COLUMN]

# Encode target if it is text (e.g., strings like 'Blight', 'Rust')
# We initialize le as None to handle cases where y is already numeric
le = None
if y.dtype == "object":
    le = LabelEncoder()
    y = le.fit_transform(y)

print("Features shape:", X.shape)
print("Target shape:", y.shape)

# ------------------------------------------------------
# 4. TRAIN-TEST SPLIT
# ------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

# ------------------------------------------------------
# 5. TRAIN MODELS
# ------------------------------------------------------
print("\n--- Training Models ---")

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
rf_model.fit(X_train, y_train)

svm_model = SVC(kernel="rbf", probability=True)
svm_model.fit(X_train, y_train)

print("Models trained successfully")

# ------------------------------------------------------
# 6. EVALUATION FUNCTION
# ------------------------------------------------------
def evaluate_model(model, model_name):
    # Note: X_test and y_test are accessed from global scope
    y_pred = model.predict(X_test)

    print(f"\n📌 {model_name} Results")
    print("-" * 50)
    print(f"Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}%")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))

    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(f"Confusion Matrix - {model_name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

# ------------------------------------------------------
# 7. MODEL EVALUATION
# ------------------------------------------------------
evaluate_model(rf_model, "Random Forest")
evaluate_model(svm_model, "Support Vector Machine")

# ------------------------------------------------------
# 8. FEATURE IMPORTANCE (Random Forest)
# ------------------------------------------------------
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
}).sort_values(by="Importance", ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x="Importance", y="Feature", data=importance)
plt.title("Feature Importance (Traditional ML Features)")
plt.show()

print("\nTop 5 Important Features:")
print(importance.head())

# ------------------------------------------------------
# 9. PREDICTION FUNCTION
# ------------------------------------------------------
def predict_disease(symptom_dict):
    """
    symptom_dict example:
    { 'Yellow_Halo': 1, 'Black_Mold': 0, ... }
    """
    # Create DF
    sample = pd.DataFrame([symptom_dict])

    # CRITICAL FIX: Ensure columns match training data order
    # (If missing, fill with 0)
    for col in X.columns:
        if col not in sample.columns:
            sample[col] = 0
    sample = sample[X.columns]

    prediction = rf_model.predict(sample)[0]

    # Inverse transform only if LabelEncoder was used
    if le is not None:
        prediction = le.inverse_transform([prediction])[0]

    return prediction

print("\n✅ System Ready for Crop Disease Detection")
```
